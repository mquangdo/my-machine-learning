```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

# Loading data

```python
#loading the dataframe
mail_data = pd.read_csv('mail_data.csv')
```

```python
mail_data
```

```python
mail_data.info()
```

```python
#replace Null values in to Null string
mail_data = mail_data.fillna('')
```

```python
mail_data.head()
```

```python
#label spam as 0, ham as 1
mail_data.loc[mail_data['Category'] == 'spam', 'Category'] = 0
mail_data.loc[mail_data['Category'] == 'ham', 'Category'] = 1
```

```python
mail_data
```

```python
#seperate data by label
X = mail_data['Message']
Y = mail_data['Category']
```

```python
#splitting in to training and test data#
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state=3)
```

# Feature Extraction



```python
feature_extracion = TfidfVectorizer(min_df=1, stop_words='english', lowercase=True)
feature_extracion.fit(X_train)
X_train_features = feature_extracion.transform(X_train)
X_test_features = feature_extracion.transform(X_test)

Y_train = Y_train.astype(int)
Y_test = Y_test.astype(int)
```

```python
print(X_train_features)
```

# Training model

## Non-emsemble models

```python
from sklearn.svm import SVC
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold, GridSearchCV
from sklearn.neighbors import KNeighborsClassifier
```

```python
lg_clf = LogisticRegression()
lg_clf_param = {'penalty':['11','l2','elasticnet'], 'C':range(1, 10)}

svm_clf = SVC()
svm_clf_param = {'C':range(1, 10), 'kernel':['rbf', 'poly']}

nb_clf = MultinomialNB()
nb_clf_param = {'alpha':range(10)}

knn_clf = KNeighborsClassifier()
knn_clf_param = {'n_neighbors':range(1, 10)}

kfold = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)
```

### Logistic Regression

```python
search = GridSearchCV(estimator=lg_clf, param_grid=lg_clf_param, cv=kfold)
search.fit(X_train_features, Y_train)
search.best_params_
```

```python
lg_clf = LogisticRegression(C=9)
```

```python
lg_clf.fit(X_train_features, Y_train)
lg_clf.score(X_test_features, Y_test)
```

## Multinomial Naive Bayes

```python
search = GridSearchCV(estimator=nb_clf, param_grid=nb_clf_param, cv=kfold)
search.fit(X_train_features, Y_train)
search.best_params_
```

```python
nb_clf = MultinomialNB(alpha=1)
```

```python
nb_clf.fit(X_train_features, Y_train)
nb_clf.score(X_test_features, Y_test)
```

## Support Vectors Machine

```python
svm_clf = SVC()
```

```python
svm_clf.fit(X_train_features, Y_train)
svm_clf.score(X_test_features, Y_test)
```

## KNN

```python
search = GridSearchCV(estimator=knn_clf, param_grid=knn_clf_param, cv=kfold)
search.fit(X_train_features, Y_train)
search.best_params_
```

```python
knn_clf = KNeighborsClassifier(n_neighbors=2)
```

```python
knn_clf.fit(X_train_features, Y_train)
knn_clf.score(X_test_features, Y_test)
```
