```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# Exploratory Data Analysis

```python
df = pd.read_csv('data/train.csv')
```

```python
df.head()
```

```python
df.dtypes
```

```python
df.info()
```

```python
df_survived = df[df['Survived'] == 1]
df_dead = df[df['Survived'] == 0]
```

```python
def visualize_cat_variable(data, col, hue=None, title=None):
    sns.countplot(data, x=col, hue=hue)
    plt.title(title)
    plt.show()
```

```python
visualize_cat_variable(df, 'Sex', hue='Survived', title='Surviving stat based on sex')
```

Tỉ lệ nữ sống sót cao hơn tỉ lệ nam.

```python
visualize_cat_variable(df, 'Pclass', hue='Survived', title='Surviving stat based on sex')
```

Tỉ lệ sống sót ở các khoang cao cấp thì cao hơn khoang thông thường.

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 6))
sns.barplot(data=df, x='Pclass', y='Survived', palette=['green'], ax=ax[0])
ax[0].set_title('Survived Rate by Pclass')
sns.countplot(data=df, x='Pclass', hue='Survived', palette=['red', 'blue'], ax=ax[1])
ax[1].set_title('Survived or Dead by Pclass')
```

```python
sns.barplot(data=df, x='Pclass', y='Survived', hue='Sex', palette=['red', 'blue'])
plt.title('Survival rate by Pclass and Sex')
```

```python
df['Age'].hist()
plt.title('Age histogram')
```

```python
df_survived['Age'].hist()
plt.title('Age histogram of survived')
plt.xlabel('Age')
```

```python
df_dead['Age'].hist()
plt.title('Age histogram of dead')
plt.xlabel('Age')
```

Số người chết thường là các người từ vị thành niên đến trung niên, tương tự với số người sống.

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 6))

sns.countplot(df_survived, x='Pclass', hue='Sex', ax=ax[0])
ax[0].set_title('Survived members by class and gender')
sns.countplot(df_dead, x='Pclass', hue='Sex', ax=ax[1])
ax[1].set_title('Dead members by class and gender')
```

```python
plt.scatter(df['Fare'], df['Pclass'])
plt.title('Relationship between fare and pclass')
plt.ylabel('Pclass')
plt.xlabel('Fare')
```
